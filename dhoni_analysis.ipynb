```python
import json
import os
import pandas as pd

# Point this to wherever you saved the cricsheet folder
folder_path = "cricket_data"

# Load all match files
all_matches = []
for filename in os.listdir(folder_path):
    if filename.endswith(".json"):
        with open(os.path.join(folder_path, filename)) as f:
            all_matches.append(json.load(f))

print(f"Total matches loaded: {len(all_matches)}")
```

```python
# See just the top level keys
match = all_matches[0]
print(match.keys())
```

```python
# See match info
info = match['info'].keys()
print(info)
```

```python
print("Teams:", match['info']['teams'])
print("Outcome:", match['info']['outcome'])
print("Player of match:", match['info']['player_of_match'])
print("Event:", match['info']['event'])
print("Date:", match['info']['dates'])
```

```python
# Find all matches where Dhoni was in the squad
dhoni_matches = []

for match in all_matches:
    players = match['info'].get('players', {})
    # Flatten all players from both teams
    all_players = [p for team in players.values() for p in team]
    if 'MS Dhoni' in all_players:
        dhoni_matches.append(match)

print(f"Matches with Dhoni: {len(dhoni_matches)}")
```

```python
# India's win/loss record in Dhoni matches
wins = 0
losses = 0
no_result = 0

for match in dhoni_matches:
    outcome = match['info'].get('outcome', {})
    winner = outcome.get('winner', None)
    
    if winner == 'India':
        wins += 1
    elif winner is not None:
        losses += 1
    else:
        no_result += 1

print(f"Wins: {wins}")
print(f"Losses: {losses}")
print(f"No result/ties: {no_result}")
print(f"Win rate: {round(wins/(wins+losses)*100, 1)}%")
```

```python
from datetime import datetime

dhoni_captain_matches = []

for match in dhoni_matches:
    date_str = match['info']['dates'][0]
    year = int(date_str.split('-')[0])
    
    if 2007 <= year <= 2014:
        dhoni_captain_matches.append(match)

print(f"Dhoni captaincy era matches: {len(dhoni_captain_matches)}")
```

```python
from collections import Counter

teammate_counter = Counter()

for match in dhoni_captain_matches:
    players = match['info'].get('players', {})
    india_players = players.get('India', [])
    for player in india_players:
        if player != 'MS Dhoni':
            teammate_counter[player] += 1

for player, count in teammate_counter.most_common(15):
    print(f"{player}: {count} matches")
```

```python
chasing_wins, chasing_losses = 0, 0
defending_wins, defending_losses = 0, 0

for match in dhoni_captain_matches:
    toss = match['info'].get('toss', {})
    outcome = match['info'].get('outcome', {})
    winner = outcome.get('winner', None)
    
    if winner is None:
        continue
    
    toss_winner = toss.get('winner')
    toss_decision = toss.get('decision')
    
    # Figure out if India was chasing or defending
    if toss_winner == 'India':
        india_chasing = toss_decision == 'field'
    else:
        india_chasing = toss_decision == 'bat'
    
    is_win = winner == 'India'
    
    if india_chasing:
        if is_win: chasing_wins += 1
        else: chasing_losses += 1
    else:
        if is_win: defending_wins += 1
        else: defending_losses += 1

chasing_rate = round(chasing_wins/(chasing_wins+chasing_losses)*100, 1)
defending_rate = round(defending_wins/(defending_wins+defending_losses)*100, 1)

print(f"Chasing — Wins: {chasing_wins}, Losses: {chasing_losses}, Win rate: {chasing_rate}%")
print(f"Defending — Wins: {defending_wins}, Losses: {defending_losses}, Win rate: {defending_rate}%")
```

```python
def get_dhoni_batting_position(match):
    for innings in match['innings']:
        if innings['team'] == 'India':
            batting_order = []
            for over in innings['overs']:
                for delivery in over['deliveries']:
                    batter = delivery['batter']
                    if batter not in batting_order:
                        batting_order.append(batter)
            if 'MS Dhoni' in batting_order:
                return batting_order.index('MS Dhoni') + 1  # +1 to make it 1-based
    return None

# Test on a few matches
for match in dhoni_captain_matches[:5]:
    pos = get_dhoni_batting_position(match)
    date = match['info']['dates'][0]
    teams = match['info']['teams']
    print(f"{date} | {teams} | Dhoni batted at: {pos}")
```

```python
early = {'wins': 0, 'losses': 0}  # positions 1-5
late = {'wins': 0, 'losses': 0}   # positions 6+

for match in dhoni_captain_matches:
    pos = get_dhoni_batting_position(match)
    outcome = match['info'].get('outcome', {})
    winner = outcome.get('winner', None)
    
    if pos is None or winner is None:
        continue
    
    is_win = winner == 'India'
    
    if pos <= 5:
        if is_win: early['wins'] += 1
        else: early['losses'] += 1
    else:
        if is_win: late['wins'] += 1
        else: late['losses'] += 1

early_rate = round(early['wins'] / (early['wins'] + early['losses']) * 100, 1)
late_rate = round(late['wins'] / (late['wins'] + late['losses']) * 100, 1)

print(f"Dhoni batted early (pos 1-5): {early_rate}% ({early['wins']}W {early['losses']}L)")
print(f"Dhoni batted late (pos 6+):   {late_rate}% ({late['wins']}W {late['losses']}L)")
```
